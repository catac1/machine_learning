```python
import pandas as pd

customer_master = pd.read_csv("http://114.207.245.181:13000/csv/customer_master.csv")
customer_master.to_csv("./csv/customer_master.csv")
```

```python
customer_master.info()
```

```python
customer_master.columns
```

```python
# 딕셔너리 타입으로 앞쪽에 원래:변경
t1 = {
    'customer_id': '고객아이디',
    'customer_name': '고객명',
    'registration_date': '등록일시',
    'email': '이메일',
    'gender': '성별',
    'age': '나이',
    'birth': '생년월일',
    'pref': '지역',
}

# inplace 옵션이 기본 False 이기때문에 생략하면 바뀌지 않는다
# customer_master = customer_master.rename(columns=t1, inplace=False)
customer_master = customer_master.rename(columns=t1)
customer_master
```

```python
customer_master.dtypes
```

```python
# 현재 등록일시의 타입이 문자인데 날짜 타입으로 변환하기
customer_master['등록일시'] = pd.to_datetime(customer_master['등록일시'], format='%Y-%m-%d %H:%M')
customer_master['생년월일'] = pd.to_datetime(customer_master['생년월일'], format='%Y-%m-%d')
```

```python
customer_master['년'] = customer_master['생년월일'].dt.year
customer_master['월'] = customer_master['생년월일'].dt.month
customer_master['일'] = customer_master['생년월일'].dt.day
```

```python
# 나이 다시 계산하기 한국식 나이이므로 +1
customer_master['나이'] = 2026 - customer_master['년'] + 1
customer_master
```

```python
bins = [0, 9, 19, 29, 39, 49, 59, 69, 79, 150]
labels = []
labels = ["0대", "10대", "20대", "30대", "40대", "50대", "60대", "70대", "80이상",]
customer_master['연령대'] = pd.cut(customer_master['나이'], bins=bins, labels=labels, right=True)
customer_master
```

```python
customer_master.to_pickle('./pkl/customer_master.pkl')
```

```python
host = 'http://114.207.245.181:13000'
transaction_1 = pd.read_csv(f'{host}/csv/transaction_1.csv')
transaction_1.head(3)
```

```python
transaction_2 = pd.read_csv(f'{host}/csv/transaction_2.csv')
transaction_2.head(3)
```

```python
# transaction_1에 transaction_2를 붙이기

transaction = pd.concat([transaction_1, transaction_2], ignore_index=True)
print(transaction.columns)
transaction
```

```python
t3 = dict(zip(
    ['transaction_id', 'price', 'payment_date', 'customer_id'], 
    ['거래코드', '결제금액', '결제일시', '고객아이디',]
))
transaction = transaction.rename(columns = t3)
transaction['결제일시'] = pd.to_datetime(transaction['결제일시'], format='%Y-%m-%d %H:%M:%S')
print(transaction.dtypes)
transaction.to_pickle('./pkl/transaction_master.pkl')
transaction
```
