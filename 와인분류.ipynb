```python
kor_names = [
    "와인종류",
    "고정산도",
    "휘발성산도",
    "구연산",
    "잔당",
    "염화물",
    "유리 이산화황",
    "총 이산화황",
    "밀도",
    "산성도 지수",
    "황산염",
    "알코올도수",
    "품질등급",
]
t1 = dict(zip(df.columns, kor_names))
t1
```

```python
df.rename(columns=t1, inplace=True)
```

```python
df['와인종류'] = df['와인종류'].map({'white': 0, 'red': 1})
```

```python
df.dtypes
```

```python
df1 = df[
    ["와인종류",
    "고정산도",
    "휘발성산도",
    "구연산",
    "잔당",
    "염화물",
    "유리 이산화황",
    "총 이산화황",
    "밀도",
    "산성도 지수",
    "황산염",
    "알코올도수",
]]
```

```python
df1.columns
```

```python
df1.dtypes
```

```python
y = df1['와인종류'].values
x = df1.drop(['와인종류'], axis=1).values
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=456, stratify=y)

x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
#평균을 0 편차를 1
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
x_train_scaled.shape, x_test_scaled.shape
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train_scaled, y_train)

lr.score(x_train_scaled, y_train), lr.score(x_test_scaled, y_test)
```

```python
y_pred = lr.predict(x_test_scaled)
y_pred[:10], y_test[:10]
```

```python
x_columns = [
    "고정산도",
    "휘발성산도",
    "구연산",
    # "잔당",
    "염화물",
    # "유리 이산화황",
    # "총 이산화황",
    "밀도",
    # "산성도 지수",
    "황산염",
    # "알코올도수"
]
```

```python
df2 = df[x_columns]
df3 = df[['품질등급']]
```

```python
df3['품질등급'].replace({9: 6, 3: 5, 4:5, 8:6}, inplace=True)
```

```python
df3['품질등급'].value_counts()
```

```python
y_qual = df3['품질등급'].values
x_qual = df2.values
print(x_qual.shape, y_qual.shape)
from sklearn.model_selection import train_test_split

x_qual_train, x_qual_test, y_qual_train, y_qual_test = train_test_split(
    x_qual, 
    y_qual, 
    test_size=0.2, 
    # random_state=456,
    # stratify=y
)

print(x_qual_train.shape, x_qual_test.shape, y_qual_train.shape, y_qual_test.shape)
#평균을 0 편차를 1
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_qual_train_scaled = scaler.fit_transform(x_qual_train)
x_qual_test_scaled = scaler.transform(x_qual_test)
x_qual_train_scaled.shape, x_qual_test_scaled.shape
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=800,
    random_state=123,
    max_depth=11,
)

rf.fit(x_qual_train_scaled, y_qual_train)
print(rf.score(x_qual_train_scaled, y_qual_train), rf.score(x_qual_test_scaled, y_qual_test))
y_qual_pred = rf.predict(x_qual_test_scaled)
y_qual_pred[:10], y_qual_test[:10]
```

```python
depths = [tree.tree_.max_depth for tree in rf.estimators_]

print("평균 깊이:", sum(depths)/len(depths))
print("최대 깊이:", max(depths))
```

```python

importance = pd.Series(
    rf.feature_importances_,
    index=x_columns,
)

print(importance.sort_values(ascending=False))
```
