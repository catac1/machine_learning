```python
import pandas as pd

# 가져올 collection 선택
c1 = db.get_collection("customer_master")
c2 = db.get_collection("item_master")
c3 = db.get_collection("transaction_master")

# 전체 데이터를 가져오는데 _id정보는 제거해서
d1 = c1.find({}, {"_id": 0})
d2 = c2.find({}, {"_id": 0})
d3 = c3.find({}, {"_id": 0})

# 가져온 데이터를 DataFrame으로 변경
customer = pd.DataFrame(list(d1))
item = pd.DataFrame(list(d2))
transaction = pd.DataFrame(list(d3))
```

```python
transaction_d1 = pd.read_csv('http://114.207.245.181:13000/csv/transaction_detail_1.csv')
transaction_d2 = pd.read_csv('http://114.207.245.181:13000/csv/transaction_detail_2.csv')

transaction_detail = pd.concat([transaction_d1, transaction_d2], ignore_index=True)
transaction_detail.info()
```

```python
transaction_detail.columns.to_list()
```

```python
d_for_rename = dict(zip(
    ['detail_id', 'transaction_id', 'item_id', 'quantity'],
    ["거래번호", "거래코드", "물품코드", "수량"]
))
```

```python
transaction_detail = transaction_detail.rename(columns=d_for_rename)
transaction_detail
```

```python
# transaction: 중국집에 54000, transaction_detail: 짜장1개, 짬뽕2 등
print(transaction.shape, transaction_detail.shape)
```

```python
transaction_data = pd.merge(
    transaction[['거래코드', '결제일시', '고객아이디']],
    transaction_detail[['거래코드', '물품코드', '수량']],
    on="거래코드",
    how="right"
)

transaction_data.info()
```

```python
transaction_data.head(3)
```

```python
item
```

```python
customer.shape
```

```python
transaction_item_data = pd.merge(
    transaction_data,
    item,
    left_on="물품코드",
    right_on="물품코드",
    how="left"
)
transaction_item_data
```

```python
customer.head(1)
```

```python
data =pd.merge(
    transaction_item_data,
    customer,
    on="고객아이디",
    how="left"
)
print(data.shape)
data.head(3)
```

```python
# 거래금액 생성
data['거래금액'] = data['수량'] * data['물품가격']
```
