```python
import numpy as np
from scipy.special import softmax

# 1. 단어 백터 생성
words = { 
    "나는" : np.array([1.0, 0.0]),
    "사과를" : np.array([1.0, 1.0]),
    "먹는다" : np.array([0.0, 1.0]),
}

# 2. 문장을 행렬로 변경
X = np.array(list(words.values()))

# Q : 찾고 싶은 정보, K : 다른 단어들의 특징, V : 실제 전달할 내용
Q = X
K = X
V = X

print("\nQ=\n")
print(Q)

print("\nK=\n")
print(K.T)

print("\nQ*K.T=\n")
print(np.dot(Q, K.T))


attention_weights = softmax(np.dot(Q, K.T), axis=1)
print("\nattention_weights=\n")
attention_weights


# 단어 자체의 의미를 주변 문맥을 반영한 의미로 변환됨
output = np.dot(attention_weights, V)
print("\noutput=\n")
print(output)
```

```python
# 토크나이즈 허깅페이스에서 다운로드하기
from transformers import AutoTokenizer

model_name = "klue/bert-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 문장을 단어로 구분하기
text = "나는 사과를 좋아한다"
tokenizer.tokenize(text)
```

```python
# 문자를 숫자로 변경
ids = tokenizer.encode(text)
ids
```

```python
texts = [
    "오늘은 비가 옵니다",
    "이제는 비가 안오는것 같습니다",
    "내일은 날씨가 맑으면 좋겠습니다"
]

inputs = tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="np",
    max_length=100,
)

# input_ids, token_type_ids, attention_mask의 키로 구성됨
inputs
```

```python
import tensorflow as tf
input_ids = tf.convert_to_tensor(inputs["input_ids"])
attention_mask = tf.convert_to_tensor(inputs['attention_mask'])

input_ids.shape, attention_mask.shape
```
