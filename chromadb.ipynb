```python
import chromadb
from chromadb.utils import embedding_functions

# 임베딩 모델
model_name = "snunlp/KR-SBERT-V40K-klueNLI-augSTS"

embedding_func = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name = model_name
)
```

```python
# chromadb 접속

client_db = chromadb.HttpClient(
    host="127.0.0.1", port=28000
)

# 컬렉션 생성

collection = client_db.get_or_create_collection(
    name="book_collection", embedding_function=embedding_func
)

# 데이터 추가하기
# documents => 벡터검색 (유사도)
# metadatas => 조건 검색
# ids => 기본키(자동생성 없음, 자체적으로 고유하게 만들어야 함)
# metadatas 에 있는 데이터들이 일반적 데이터베이스 query문에 사용되는 것처럼 사용됨
collection.add(
    documents = [
        "파이썬 머신러닝 완벽 가이드는 머신러닝의 기초를 설명하는 책입니다.", 
        "react는 웹 화면을 만드는 것을 설명하는 책입니다."
    ],
    metadatas = [
        { 
            "title": "파이썬 머신러닝 가이드",
            "img": "./images/book1.jpg",
            "category" : "AI",
        },
        { 
            "title": "리액트 가이드",
            "img": "./images/book2.jpg",
            "category" : "WEB",
        },
    ],
    ids = ["b1", "b2"]
)
```

```python
# 조회 

collection = client_db.get_collection(
    name="book_collection", embedding_function=embedding_func
)

result = collection.query(
    query_texts=["파이썬 머신러닝"],
    where = { "category": "AI"},
    n_results=1
)

result
```

```python
import pandas as pd
dtype={13: str, 14: str, 17: str}
df = pd.read_csv("./csv/OpenData_PotOpenTabletIdntfcC20260609.csv", 
    dtype=dtype, 
    # na_values=['-', '1호', '4호', '15.50-15.60', '14.2.', '21.1±0.1mm'])
)
df.head()

df['품목명'].to_list()
```

```python
collection = client_db.get_or_create_collection(
    name="test1", embedding_function=embedding_func
)

documents = df['품목명'].to_list()
metadatas = [
    { 
        "title": row['분류번호'],
        "img": row['전문일반구분'],
    }
    for _, row in df.iterrows()
]
ids = [ str(i) for i in range(len(df))]
```

```python
batch_size = 5000
for i in range(0, len(documents), batch_size):
    collection.add(
        documents = documents[i: i + batch_size],
        metadatas = metadatas[i: i + batch_size],
        ids = ids[i: i + batch_size],
    )
    print(f"Added {i +len(documents[i:i+batch_size])} / {len(documents)}")
```

```python
result = collection.query(
    query_texts=["페라"],
    where={'img': "전문의약품"},
    n_results=5
)
result
```

```python
from openai import OpenAI

client = OpenAI()

context = "이 부분은 DB에서 조회한 최신정보로 구성"
question = "해열제에 대해서 설명해줘"

prompt = f"""
    너는 약에 대한 전문가다
    아래 문서를 참고해서 답변해

    문서:
    { context }

    질문: 
    { question }

    답변:
"""


response = client.chat.completions.create(
    model="gpt-5.5",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

print(f"답변: {response.choices[0].message.content}")
```

```python
from langchain_ollama import OllamaLLM
from crewai import LLM

ollama_llm = LLM(
    model = "ollama/exaone-3.5:latest",    
    base_url = "http://192.168.0.12:11435",
    temperature = 0.3 # 0.3은 코드 요약 분석용
)
```

```python
# Agent 설정
from crewai import Agent

researcher = Agent(
    role = "AI Researcher",
    goal = "주어진 주제에 대해서 개념을 조사한다.",
    backstory = "AI 논문과 기술 문서를 분석하는 전문가",
    llm = ollama_llm,
    verbose = True
)

writer = Agent(
    role = "Technical Writer",
    goal = "조사 결과를 개발자 중심으로 정리한다.",
    backstory = "개발자를 위한 기술 문서를 잘 쓰는 작가",
    llm = ollama_llm,
    verbose = True
)
```

```python
# Task 정의
from crewai import Task

research_task = Task(
    description = "CrewAI와 Ollama의 역할의 장점을 조사해라",
    expected_output = "CrewAI와 Ollama의 핵심 역할과 장점을 요약한 정리 문서",
    agent = researcher
)

writer_task = Task(
    description = "조사결과를 개발자용으로 쉽게 정리해라",
    expected_output = "비 전공자도 이해 할 수 있는 쉬운 설명 문서",
    agent = writer
)
```

```python
# 에이전트 + Task 묶기
from crewai import Crew

# task에 의해서 순서가 결정됨
crew = Crew(
    agents = [researcher, writer],
    tasks = [research_task, writer_task],
    verbose = True
)
```

```python
result = await crew.kickoff_async()
result
```
