```python
t1 = {"Freshman": "1학년", "Sophomore": "2학년", "Junior": "3학년", "Senior": "4학년"}
df['학년'] = df['학년'].map(t1)
df
```

```python
import pandas as pd
df1 = pd.read_csv('http://114.207.245.181:13000/csv/score_info.csv')
df1
```

```python
print(df1.columns)
t_for_column_rename = dict(zip(df1.columns.to_list(), ["학생아이디","수학", "과학", "영어", "역사", "체육"]))
```

```python
df1.rename(columns=t_for_column_rename, inplace=True)
```

```python
df1.head(3)
```

```python
df2 = pd.merge(df, df1, on="학생아이디")
df2
```

```python
df2['합계'] = df2['수학'] + df2['과학'] + df2['영어'] + df2['역사'] + df2['체육'] 
df2
```

```python
df2['평균'] = df2['합계'] / 5
df2
```

```python
# min은 예시) 1 2 2 4 => 같은 랭크일때 낮은 랭크로 취급. max는 높은랭크 => 1 3 3 4

df2['석차'] = df2['합계'].rank(method="min", ascending=False).astype(int)
df2
```

```python
# 학년별 합계 평균 구하기
# groupby([]).sum, mean, min, max, ...

df2.groupby(['학년'])[['수학', '영어']].mean().round(2)
```
